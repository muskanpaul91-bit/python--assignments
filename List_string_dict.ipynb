## 1. Reverse a List Without Using reverse()

### Problem Statement
Write a program to reverse a list without using the built-in reverse() method.

### Example Input
```python
1 2 3 4 5
```

### Example Output
```python
[5, 4, 3, 2, 1]
```


````python
l=[1,2,3,4,5]
print(l[::-1])
````

````python
l=[1,2,3,4,5]
j=0
for i in range(len(l)-1,len(l)//2,-1):
    l[i],l[j]=l[j],l[i]
    j+=1
print(l)
````

````python
l=[1,2,3,4,5]
left=0
right=len(l)-1
while left<right:
    l[left],l[right]=l[right],l[left]
    left+=1
    right-=1
print(l)
````


## 2. Find Duplicate Elements in a List

### Problem Statement
Write a program to find all duplicate elements in a list.

### Example Input
```python
1 2 3 2 4 5 1
```

### Example Output
```python
[1, 2]
```


````python
l=[1,2,3,2,4,5,1]
new=[]
duplicates=[]
for i in l:
    if i not in new:
        new.append(i)
    else:
        duplicates.append(i)
print(new,duplicates)
````

### Problem Statement
Write a program to replace all spaces in a string with hyphens (-).

### Example Input
```python

```

### Example Output
```python

```


````python
str1='data science course'
new=''
for i in str1:
    if i.isspace():
        new+='-'
    else:
        new+=i
print(new)
````


## 15. Rotate List by K Positions

### Problem Statement
Write a program to rotate a list by K positions.

### Example Input
```python
List=[1,2,3,4,5], K=2
```

### Example Output
```python
[4, 5, 1, 2, 3]
```


````python
list=[1,2,3,4,5]
k=2
for j in range(k):
    last=list[-1]
    for i in range(len(list)-1,0,-1):
        list[i]=list[i-1]
    list[0]=last
print(list)
````
