# Q1. Sum of Multiple Numbers using *args

## Problem Statement
Write a Python function using *args to calculate the sum of all numbers passed as arguments.

## Sample Function Call
```python
addition(10, 20, 30)
```

## Sample Output
```python
60
```


````python
def sum(*s):
    sum=0
    for i in s:
        sum+=i
    print(sum)
sum(10,20,30)
````

# Q2. Find Maximum Number using *args

## Problem Statement
Write a Python function using *args to find the largest number.

## Sample Function Call
```python
maximum(5, 20, 15, 50, 10)
```

## Sample Output
```python
50
```


````python
def largest(*l):
    largest=l[0]
    for i in l:
        if i>largest:
            largest=i
    print(largest)
largest(5,20,15,50,10)
````

# Q9. Multiply All Numbers using *args

## Problem Statement
Write a Python function using *args to multiply all numbers.

## Sample Function Call
```python
multiply(2, 3, 4)
```

## Sample Output
```python
24
```


````python
def multiple(*m):
    mul=1
    for i in m:
        mul*=i
    return mul
print(multiple(2,3,4))
````
