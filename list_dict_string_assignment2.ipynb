# Q1. Reverse Each Word in a String

## Problem Statement
Write a Python program to take a sentence from the user and reverse each word individually.

## Sample Input
```python
Enter sentence: hello world
```

## Sample Output
```python
olleh dlrow
```


````python
n= input("Enter your message:-")
l=n.split()
for i in l:
    print(i[::-1], end =' ')
````

# Q2. Find Frequency of Characters

## Problem Statement
Write a Python program to count the frequency of each character in a string.

## Sample Input
```python
Enter string: apple
```

## Sample Output
```python
a : 1
p : 2
l : 1
e : 1
```


````python
n="apple"
d={}
for i in n:
    if i not in d:
        d[i]=1
    else:
        d[i]+=1
print(d)
````

# Q3. Remove Duplicate Elements from List

## Problem Statement
Write a Python program to remove duplicate elements from a list.

## Sample Input
```python
Enter list: [1,2,2,3,4,4,5]
```

## Sample Output
```python
[1, 2, 3, 4, 5]
```


````python
n=[1,2,2,3,4,5,4]
duplicate=[]
for i in n:
    if i not in duplicate:
        duplicate.append(i)
print(duplicate)
````

# Q4. Find Second Largest Number in List

## Problem Statement
Write a Python program to find the second largest element in a list.

## Sample Input
```python
Enter list: [10, 20, 5, 40, 30]
```

## Sample Output
```python
30
```


````python
n=[10,20,5,40,30]
largest=n[0]
second=n[0]
for i in n:
    if i>largest:
        second=largest
        largest=i
    elif i>second and i!=largest:
        second=i
print(second)
````

# Q5. Merge Two Dictionaries

## Problem Statement
Write a Python program to merge two dictionaries into one.

## Sample Input
```python
dict1 = {'a':1, 'b':2}
dict2 = {'c':3, 'd':4}
```

## Sample Output
```python
{'a': 1, 'b': 2, 'c': 3, 'd': 4}
```


````python
dict1={"a":1, 'b':2}
dict2={'c':3, 'd':4}
for i in dict2:
    dict1[i]=dict2[i]
print(dict1)
````

# Q6. Find Common Elements Between Two Lists

```python
list1 = [1,2,3,4]

```

## Sample Output

# Q9. Create Dictionary from Two Lists

## Problem Statement
Write a Python program to create a dictionary using two lists.

## Sample Input
```python
keys = ['name', 'age']
values = ['Shiva', 22]
```

## Sample Output
```python
{'name': 'Shiva', 'age': 22}
```


````python
keys=['name','age']
values=['shiva',22]
dic={}
j=0
for i in keys:
    if i not in dic:
        dic[i]=values[j]
        j+=1
print(dic)
    
````

````python
keys=['name','age']
values=['shiva',22]
print(dict(zip(keys,values)))
````

# Q10. Find Longest Word in String

## Problem Statement
Write a Python program to find the longest word in a sentence.

## Sample Input
```python
Enter sentence: Python programming is amazing
```

## Sample Output
```python
programming
```


````python
n='python programming is amazing'
l=n.split()
longest=''
for i in l:
    if len(i)>len(longest):
        longest=i
print(longest)
````
