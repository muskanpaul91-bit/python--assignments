# Q1. Create a Student Class

## Problem Statement
Create a class named `Student` and create an object of that class.

Print:
- Student Name
- Student Age

## Sample Output
```python
Name: Raj
Age: 17
```


````python
class student:
    def na(self,Name,Age):
        self.Name=Name
        print("Name:",self.Name)
        print("Age:",Age)
obj=student()
obj.na("Raj",17)
````
